##### 3. Feature Aggregation or Transformation

The 'year' feature will be transformed to create a new feature 'age' to represent the age of the vehicle. From personal experience and as a common knowledge fact the age of a vehicle is very important in determining its value. If a vehicle is 3 years old or 6 years old it would stand to reason that that is a big difference, thus by transforming that feature from a year value to an age value (measured in years back from a "now" point) that data will be more useful. We will however also transform the 'year' feature into 'year_model', because it is also conceivable that a model of a given year could have unique value, because it was made on that specific year. Cars will often have unique features on a given year, for this reason we will actually transform the year feature into a categorical feature (after we've used it to calculate age). We will perform one hot encoding on that feature, similar to other categorical features.

The 'condition' feature will be transformed to an ordinal feature in this order:

### Data Preparation

After our initial exploration and fine tuning of the business understanding, it is time to construct our final dataset prior to modeling.  Here, we want to make sure to handle any integrity issues and cleaning, the engineering of new features, any transformations that we believe should happen (scaling, logarithms, normalization, etc.), and general preparation for modeling with `sklearn`. 

For an initial exploration see file: https://github.com/elmunoz42/used-car-inventory-optimizer/blob/main/used-car-inventory-exploration.ipynb in this repository


```python
# Reduce the data scope to California cars only.
data = data.query('state == "ca"')
data = data.drop('state', axis=1)
```

```python
# Drop some tables that are not needed:
data = data.drop(['VIN','id'], axis=1)
```

```python
# Review missing values to determine best approach to handle them.

import seaborn as sns
import matplotlib.pyplot as plt
import importlib
# From functions.py file
functions.plot_dataframe_missing_values(data, 'images/fig1.jpg')
```

```python
# For features with a more than 10% of missing values replace NaN with "unknown" since they are categorical values
data['size'] = data['size'].fillna('unknown')
data['cylinders'] = data['cylinders'].fillna('unknown')
data['condition'] = data['condition'].fillna('unknown')
data['paint_color'] = data['paint_color'].fillna('uknown')
data['drive'] = data['drive'].fillna('unknown')
data['type'] = data['type'].fillna('unknown')
```

```python
# For features with less than 10% missing data remove rows
data = data.dropna(subset=['manufacturer', 'title_status', 'odometer', 'model', 'transmission', 'year', 'fuel'])

# Reset the index
data = data.reset_index(drop=True)
```

```python
# The max value for year in the dataset is 2022 so we'll use that as the Now date

# Ensure the 'year' column is numeric
data['year'] = pd.to_numeric(data['year'], errors='coerce').astype('int64')

# Calculate the age
now = 2022
data['age'] = now - data['year']

# Transform year into categorical feature
data['year_model'] = data['year'].astype('object')

data.drop('year', axis=1, inplace=True)
data.columns
```

```python
# Transform odometer into integer
data['odometer'] = data['odometer'].astype('int64')
```

```python
# identify the numeric and categorical features
data.dtypes
```

```python
# Separate the dataset into four subsets for the different car models and the general analysis.
data_f150 = data.query('model == "f-150"').drop('model', axis=1)
data_accord = data.query('model == "accord"').drop('model', axis=1)
data_civic = data.query('model == "civic"').drop('model', axis=1)
```
